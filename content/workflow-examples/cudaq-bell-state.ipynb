# CUDA-Q Bell-state workflow

Open this notebook with `make cudaq-lab`. The code cells are tagged `skip-execution` so the portable book build can publish their source without requiring CUDA-Q on every machine. Running cells manually in JupyterLab is unaffected by that tag.

```python
import cudaq

cudaq.set_target("qpp-cpu")


@cudaq.kernel
def bell_state():
    qubits = cudaq.qvector(2)
    h(qubits[0])
    x.ctrl(qubits[0], qubits[1])
    mz(qubits)
```

```python
cudaq.set_random_seed(7)
cudaq.sample(bell_state, shots_count=1024)
```
