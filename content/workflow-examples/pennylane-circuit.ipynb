# PennyLane circuit workflow

This notebook demonstrates the normal PennyLane execution path used by the production build.

```python
import pennylane as qml
```

```python
device = qml.device("default.qubit", wires=2)


@qml.qnode(device)
def bell_probabilities():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.probs(wires=[0, 1])
```

```python
bell_probabilities()
```
