# Qiskit Bell-state workflow

This minimal notebook confirms that a Qiskit circuit executes with the locked Qiskit Aer simulator.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
```

```python
circuit = QuantumCircuit(2)
circuit.h(0)
circuit.cx(0, 1)
circuit.measure_all()
circuit.draw(output="text")
```

```python
simulator = AerSimulator()
job = simulator.run(circuit, shots=1024, seed_simulator=7)
job.result().get_counts()
```
